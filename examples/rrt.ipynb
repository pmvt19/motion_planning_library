```python
import matplotlib.pyplot as plt

from motion_planning.rrt import RRT
from motion_planning.space import PointRobot
from motion_planning.obstacle_sets import RandomSamplePassage
```

# Setup Robot & Environment

```python
env = PointRobot()
env.set_obstacles(RandomSamplePassage())
```

# Create Task to Solve

```python
start, target = env.sample_task()
```

# Initialize RRT and Search for a Path

```python
rrt = RRT(env)
path = rrt.search(start=start, target=target, max_steps=10000, goal_bias=0.1)
```

```python
env.draw_environment(plt.gca())
rrt.draw_tree(plt.gca(), path=path, show_task=True)
plt.show()
```
