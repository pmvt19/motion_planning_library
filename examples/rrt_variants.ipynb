```python
import matplotlib.pyplot as plt

from motion_planning.obstacle_sets import RandomSamplePassage
from motion_planning.search import BiDirectionalRRT, RRTStar
from motion_planning.space import PolygonalRobot
```

# Setup Base Robot & Environment

```python
env = PolygonalRobot()
env.set_obstacles(RandomSamplePassage(num_walls=3, gap_width=2))
```

```python
start, target = env.sample_task()
```

# BiDirectional RRT

```python
rrt = BiDirectionalRRT(env)
path = rrt.search(start=start, target=target, max_steps=10000, goal_bias=0.1)
```

```python
env.draw_environment(plt.gca())
rrt.draw_tree(plt.gca(), path=path, show_task=True)
plt.show()
```

# RRT*

```python
rrt_star = RRTStar(env)
path = rrt_star.search(start=start, target=target, max_steps=5000, goal_bias=0.1)
```

```python
env.draw_environment(plt.gca())
rrt_star.draw_tree(plt.gca(), path=path, show_task=True)
plt.show()
```
