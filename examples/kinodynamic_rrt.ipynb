```python
import numpy as np
import matplotlib.pyplot as plt

from motion_planning.kinodynamic_rrt import KinodynamicRRT
from motion_planning.space import DubinsCar
from motion_planning.obstacle_sets import ParkingSpace
```

# Setup Base Robot & Environment

```python
env = DubinsCar()
```

# Create Task to Solve

```python
start = env.make_state(np.array([3.0, 2.75, 0, 0, -np.pi/2]))
target = env.make_state(np.array([-4.5, -4.75, 0, 0, -np.pi/2]))
```

# Initialize Kinodynamic RRT and Search for a Path

```python
rrt = KinodynamicRRT(env, goal_radius=4, max_time_horizon=0.5)
path = rrt.search(start, target, max_steps=10000, goal_bias=0.05)
```

# Visualize RRT Tree in the Environment

```python
rrt.draw_tree(plt.gca(), path=path)
plt.show()
```
